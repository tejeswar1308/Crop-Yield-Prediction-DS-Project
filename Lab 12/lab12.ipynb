```python
import pandas as pd
from sklearn.datasets import load_iris
from statistics import mode

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of the Iris dataset:")
print(df.head())

mean_values = df.mean()
print("\nMean values:")
print(mean_values)

median_values = df.median()
print("\nMedian values:")
print(median_values)

mode_values = df.mode()
print("\nMode values:")
print(mode_values)
```

```python
import pandas as pd
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of dataset:")
print(df.head())
range_values = df.max() - df.min()
print("\nRange Values:")
print(range_values)

variance_values = df.var()
print("\nVariance values:")
print(variance_values)

std_values = df.std()
print("\nStandard Deviation Values:")
print(std_values)

Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)
IQR = Q3-Q1
print("\nInterquartile Range Values:")
print(IQR)
```

```python
import pandas as pd
import seaborn as sns

tips = sns.load_dataset("tips")

df = pd.DataFrame(tips)

print("First 5 rows of dataset:")

print(df.head())

# Select numerical columns
numeric_df = df.select_dtypes(include='number')

range_values = numeric_df.max() - numeric_df.min()

print("\nRange Values:")

print(range_values)

variance_values = numeric_df.var()

print("\nVariance values:")

print(variance_values)

std_values = numeric_df.std()

print("\nStandard Deviation Values:")

print(std_values)

Q1 = numeric_df.quantile(0.25)

Q3 = numeric_df.quantile(0.75)

IQR = Q3 - Q1

print("\nInterquartile Range Values:")

print(IQR)
```

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
print("First 5 rows of the Iris dataset:")
print(df.head())

skewness_values = df.skew()
print("\nSkewness values:")
print(skewness_values)

kurtosis_values = df.kurtosis()
print("\nKurtosis values:") 
print(kurtosis_values)

```

```python
import pandas as pd
import seaborn as sns

tips = sns.load_dataset("tips")

df = pd.DataFrame(tips)

print("First 5 rows of the Tips dataset:")

print(df.head())

# Select only numerical columns
numeric_df = df.select_dtypes(include='number')

skewness_values = numeric_df.skew()

print("\nSkewness values:")

print(skewness_values)

kurtosis_values = numeric_df.kurtosis()

print("\nKurtosis values:")

print(kurtosis_values)
```
